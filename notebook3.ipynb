```python
import pandas as pd
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("f1_dnf.csv")

df.dropna(axis=1,how='any')
df.replace('\\N', pd.NA, inplace=True)


df.dropna(axis=1, inplace=True)



df.head()
```

```python



dnf_ano = df[df['target_finish'] == 0].groupby('year').size()
finish_ano = df[df['target_finish'] == 1].groupby('year').size()


fig, ax1 = plt.subplots(figsize=(12,6))


ax1.plot(dnf_ano.index, dnf_ano.values, color='red', label='DNF', marker='o')
ax1.set_xlabel("Ano",fontsize=14, fontweight='bold')
ax1.set_ylabel("Quantidade de DNF", color='red',fontsize=14, fontweight='bold')
ax1.tick_params(axis='y', labelcolor='red')

ax2 = ax1.twinx() 
ax2.plot(finish_ano.index, finish_ano.values, color='green', marker='o', label='Corridas Terminadas')
ax2.set_ylabel("Corridas Terminadas", color='green',fontsize=14, fontweight='bold')
ax2.tick_params(axis='y', labelcolor='green')


plt.title("Comparação: DNF x Corridas Terminadas por Ano",fontsize=14, fontweight='bold')
fig.tight_layout()  


plt.show()
```
